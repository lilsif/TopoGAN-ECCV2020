```python
import os
import random
from PIL import Image

# Paths
input_dir = r"C:\TopoGAN-ECCV2020\Data\cremi\processed"
output_dir = r"C:\TopoGAN-ECCV2020\Data\cremi\64x64"

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Parameters
crop_size = 64
num_samples = 7500

# Collect all PNG files
all_png_files = [f for f in os.listdir(input_dir) if f.lower().endswith('.png')]
if len(all_png_files) == 0:
    raise ValueError("No PNG files found in the input directory.")

# Randomly sample with replacement if not enough distinct files
sampled_files = random.choices(all_png_files, k=num_samples)

# Crop and save
for i, fname in enumerate(sampled_files):
    img_path = os.path.join(input_dir, fname)
    img = Image.open(img_path).convert("L")  # convert to grayscale if needed

    if img.width < crop_size or img.height < crop_size:
        raise ValueError(f"Image {fname} is smaller than the crop size of {crop_size}x{crop_size}")

    # Random crop
    left = random.randint(0, img.width - crop_size)
    top = random.randint(0, img.height - crop_size)
    crop = img.crop((left, top, left + crop_size, top + crop_size))

    # Save cropped image
    crop.save(os.path.join(output_dir, f"crop_{i:05d}.png"))

print("Done! 7500 cropped images saved.")
```

```python
import os
import random
from PIL import Image

# Parameters
initial_crop_size = 256  # crop this size first
final_size = 64          # then downscale to this
num_samples = 7500

# Paths
input_dir = r"C:\TopoGAN-ECCV2020\Data\cremi\processed"
output_dir = fr"C:\TopoGAN-ECCV2020\Data\cremi\{final_size}x{final_size}_downscaled_from_{initial_crop_size}"

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)


# Collect all PNG files
all_png_files = [f for f in os.listdir(input_dir) if f.lower().endswith('.png')]
if len(all_png_files) == 0:
    raise ValueError("No PNG files found in the input directory.")

# Randomly sample with replacement if not enough distinct files
sampled_files = random.choices(all_png_files, k=num_samples)

# Crop, downscale, and save
for i, fname in enumerate(sampled_files):
    img_path = os.path.join(input_dir, fname)
    img = Image.open(img_path).convert("L")  # convert to grayscale if needed

    if img.width < initial_crop_size or img.height < initial_crop_size:
        raise ValueError(f"Image {fname} is smaller than the crop size of {initial_crop_size}x{initial_crop_size}")

    # Random crop of 256x256
    left = random.randint(0, img.width - initial_crop_size)
    top = random.randint(0, img.height - initial_crop_size)
    crop = img.crop((left, top, left + initial_crop_size, top + initial_crop_size))

    # Downscale to 64x64 using antialiasing
    resized = crop.resize((final_size, final_size), resample=Image.LANCZOS)

    # Save final image
    resized.save(os.path.join(output_dir, f"crop_{i:05d}.png"))

print(f"Done! {num_samples} cropped and downscaled images saved.")
```
