```python
import h5py
import numpy as np
from PIL import Image
import os

def hdf5_to_png(hdf5_path, output_dir, dataset_name):
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Open the HDF5 file (with proper path handling)
    with h5py.File(hdf5_path, 'r') as f:
        print("Datasets in the file:", list(f.keys()))
        
        volumes = f["volumes"] 
        raw_data = volumes["raw"]
        
        data = raw_data[()]  # Load data

        # Grayscale images [N, H, W]
        if data.ndim == 3:
            for i, img in enumerate(data):
                img_pil = Image.fromarray(img.astype(np.uint8))
                img_pil.save(os.path.join(output_dir, f"{dataset_name}_{i}.png"))

        # RGB images [N, H, W, 3]
        elif data.ndim == 4:
            for i, img in enumerate(data):
                img_pil = Image.fromarray(img.astype(np.uint8))
                img_pil.save(os.path.join(output_dir, f"{dataset_name}_{i}.png"))

        else:
            print(f"Skipping {dataset_name} (not 3D/4D image data)")

#     except Exception as e:
#         print(f"Error reading {hdf5_path}: {e}")

# Example usage (with corrected paths)
hdf5_path = r"C:\TopoGAN-ECCV2020\Data\cremi\raw\sample_A_20160501.hdf"
output_dir = r"C:\TopoGAN-ECCV2020\Data\cremi\processed"
dataset_name = "cremi"
hdf5_to_png(hdf5_path, output_dir,dataset_name)
```
